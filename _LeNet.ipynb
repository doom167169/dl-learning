LeNet-5 由两个部分组成：卷积编码器和全连接层密集块

```python
import torch
from torch import nn
from d2l import torch as d2l

class Reshape(torch.nn.Module):
    def forward(self, x):
        return x.view(-1, 1, 28, 28)

net = torch.nn.Sequential(
    Reshape(), nn.Conv2d(1, 6, kernel_size=5, padding=2), nn.Sigmoid(),  # padding视为了保持特征，避免削减过快
    nn.AvgPool2d(2, stride=2),  # 核大小 = 步长 → 无重叠  尺寸减半
    nn.Conv2d(6, 16, kernel_size=5), nn.Sigmoid(),
    nn.AvgPool2d(kernel_size=2, stride=2), nn.Flatten(),
    nn.Linear(16 * 5 * 5, 120), nn.Sigmoid(),  # 上一层池化后输出为16个通道，5 * 5是池化核的尺寸，展平后为16 * 5 * 5
    nn.Linear(120, 84), nn.Sigmoid(),
    nn.Linear(84, 10)  #10是类别数
)
```

```python
X = torch.rand(size=(1, 1, 28, 28), dtype=torch.float32)
for layer in net:
    X = layer(X)
    print(layer.__class__.__name__, 'output shape: \t', X.shape)
```
